## Markdown cells

Jupyter Notebooks are made from *cells*.

Your are reading the text in a *Markdown* cell.

We can put text, images, tables and [web hyperlinks](https://jupyter-notebook.readthedocs.io/en/stable/examples/Notebook/Notebook%20Basics.html) inside a Markdown cell.

A Markdown cell does NOT run code.

Everything you have read in this book so far is inside of a Markdown cell.

## Code cells

The cell below is a code cell.

As you would expect, this cell runs code. We will now see our first bit of Python code!

*Note*: you can press `Shift + Enter` on your keyboard to run a code cell:

````python
2 + 2
````

What happened here? 

When I (or you!) pressed `Shift + Enter` Jupyter Notebook sent the input (`2 + 2`) to the Python compiler (also called an interpreter).

The compiler added 2 and 2 together, and showed the output.

Here is some more, simple Python code:

````python
print("Hello world!")
````

What happened here? 

When I (or you!) pressed `Shift + Enter` my Jupyter Notebook sent the input - this time the input is text, not numbers - to the Python interpreter.

The text is inside the `print()` function. All this does is takes input and...prints it out, as you probably guessed!

We will talk more lateer about text and numbers in Python, for now, try changing the text in the cell below to make it say something you want it to say:

*Note*: one Python rule is that you have to put text in quotes, `"` or `'`. You will get an error message if you do not.

````python
print("Change this text!")
````

We have seen how to run code in code cells.

We can also write *comments* inside code cells. A comment is **NOT** sent to the Python compiler. This means it is **NOT** run as Python code. To make a comment, we put the `#` symbol at the start of the line:

````python
# This is a comment. It is not run as Python code, even when we press `Shift + Enter`
````

So, when you run a cell which only has comments inside...nothing happens!

Comments are useful for *humans*. They explain what the code does:

````python
# This is a comment. It is not run as Python code, even when we press `Shift + Enter`
# This comment explains what the code does. It prints out "Hello class!"
print("Hello class!")
````
