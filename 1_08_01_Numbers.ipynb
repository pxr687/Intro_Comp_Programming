[Dr Juan Klopper](https://www.youtube.com/c/drjuanklopper) says that programming language are really just fancy calculators. As a result, in Python, numbers work just like they do when we use a calculator.

As we have seen, we can use all the normal mathematical operations:

| Expression Type | Operator | Example    | Value     |
|-----------------|----------|------------|-----------|
| Addition        | `+`      | `2 + 3`    | `5`       |
| Subtraction     | `-`      | `2 - 3`    | `-1`      |
| Multiplication  | `*`      | `2 * 3`    | `6`       |
| Division        | `/`      | `100 / 10` | `10`      |
| Remainder       | `%`      | `101 % 10` | `1`       |
| Exponentiation (Power)  | `**`     | `10 ** 2`  | `100`     |

We also know that the two most important types of number in Python are `int` values and `float` values. The type of a number can be seen from the way it is displayed: `int` values have no decimal point and `float` values always have a decimal point.

When you put `int` and/or `float` values together with mathematical operators, it is important to know what type of data you will get back. The rules for this are shown on this page.

Here are some examples of `int` values (e.g. *counting numbers* with no decimal place). `int` values can be small (near to 0), positive (bigger than 0), negative (smaller than 0):

````python
# A small integer (`int`) near to 0, but positive.
2
````

````python
# A small integer (`int`) near to 0, but negative.
-2
````

````python
# Two integers added together, returns an `int`.
1 + 3
````

````python
# Another negative `int`.
-1
````

````python
# A negative `int` further from 0.
-10
````

`int` values can be very big in the positive direction (e.g. much bigger than 0):

````python
1234567890000000000
````

...and they can also be very "big" in the negative direction (e.g. very far from zero in the negative direction - this is actually a very small number, because it is a lot less than 0):

````python
-1234567890000000000
````

To compare to `int` values, here are some `float` values (e.g. numbers with a decimal place):

````python
# A small `float` near to 0, but positive.
1.2
````

````python
# A `float` and `int` added together (the result is a `float`).
1.5 + 2
````

````python
# Another small, positive `float`.
0.231
````

````python
# A small, negative `float`.
-0.231
````

````python
# A large, positive `float`.
184828429.01923
````

````python
# A negative `float, far from 0.
-184828429.01923
````

When a `float` value is put together with an `int` value using a mathematical
operator, then the result is always a `float` value:

````python
# float + int = float
7.5 + 4
````

````python
# int * float = float
70 * 102.3
````

Importantly, even if the decimal of a number is 0, the number is still a `float`:

````python
# A number with a decimal of 0 is still a `float`!
# float + int = float
8.0 + 2
````

Most of the time, two integers put together will make another integer, but any number (`int` or `float`)
divided by another will be a `float` value: 

````python
# Add two integers.
6 + 4
````

````python
# Divide two integers (this returns a `float`).
100/10
````

The information from this page is summarised in the table below:

| Expression Type | Operator | Data Type of Result |
|---|---|---|
| Addition | `+` | `int` if both operands are `int`; otherwise `float` |
| Subtraction | `-` | `int` if both operands are `int`; otherwise `float` |
| Multiplication | `*` | `int` if both operands are `int`; otherwise `float` |
| Division | `/` | Always `float` |
| Remainder | `%` | `int` if both operands are `int`; otherwise `float` |
| Exponentiation (Power) | `**` | `int` if both operands are `int`; otherwise `float` |

Remember that the symbols (`+`, `-`, `*`, `/`, `%`, `**`) are *operators* and the numbers we use them with are called *operands*.
