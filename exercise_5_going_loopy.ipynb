### Question 6

Now, write a card guessing game. Here is what it should do:

- Create a list called `suits` containing the strings `"hearts"`, `"diamonds"`, `"clubs"` and `"spades"`.
- Use `np.random.choice()` to randomly select one suit and store it in a variable called `secret_suit`.
- Use a `while` loop that repeats until the player guesses the correct suit.
- Inside the loop, ask the player to guess a suit using `input()`.
- Use the `.lower()` string method to convert the player's guess to lowercase.
- If the guess is incorrect, print `"Wrong suit! Try again."`.
- If the guess is correct, print `"You guessed it! Well done!"` and use `break` to stop the loop.

````python
# Your code here.
import random

suits = ["hearts", "diamonds", "clubs", "spades"]
secret_suit = random.choice(suits)

while True:
    guess = input("Guess a card suit: ").lower()

    if guess != secret_suit:
        print("Wrong suit! Try again.")
    else:
        print("You guessed it! Well done!")
        break
````
