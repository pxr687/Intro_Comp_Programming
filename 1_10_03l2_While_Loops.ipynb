# `while` loops 

`for` loops are one way we can do something lots and lots of time.

We know *exactly* how many times a `for` loop will do something ... because we tell it how many times to do something!.

For example, we know the loop in the cell below will repeat 10 times:

````python
import numpy as np
from time import sleep # We will use this later on.

n_times = 10

for i in np.arange(n_times):
    
    # We know this `print()` statement will repeat 10 times.
    print("Count me!")
````

In contrast, `while` loops will repeat some code *until a condition is `False`*.

The syntax for a `while` loop is very similar to a `for` loop.

We start with the keyword `while`, then we put an expression with a comparison operator in it (`<`, `>`, `<=`, `>=`, `==`, `!=`). We then end the header of the loop with a colon (`:`).

We can then put code in the *body* of the `while` loop by indenting, just like with a `for` loop. It is the indented code in the body that will be repeated every time the loop runs, just like with a `for` loop. 

So, the structure is as follows - the first line is the `while` loop header, the second, indented line is the body of the `while` loop:

```
while <condition>:
    <code that will be repeated>
```

A good way to understand this is to think of a countdown to a bomb exploding. (Thanks to [here](https://zvibaratz.github.io/python_for_neuroscientists/classes/class_1/class_1.html#while-loop) for this example).

This countdown is done in the cell below.

First, the `n_times` variable is set to 10.

Then the `while` loop header is made, followed by the condition `n_times > 0`. 

So, *while* `n_times` has a value bigger than 0, the loop will repeat the code in the body.

The code in the body prints the value of `n_times`, and also it prints out whether the condition (`n_times > 0`) is `True` or `False` (this is printed off to the right hand side, to make things easier to read). 

Each repeat of the loop also subtracts 1 from `n_times`. Finally, at the end of each repeat the `sleep()` function adds a small pause to make the countdown slower and easier to see as it happens.

When the loop finishes, the text `'BOOM!'` is printed - *note*: this print statement is outside of the `while` loop, so it only runs after the `while` loop is over:

````python
n_times = 10

# This is the header of the `while` loop. 
while n_times > 0:
    
    # This is the indented body of the `while` loop.
    print(f"{n_times}                                               `n_times > 0` = {n_times > 0}")
                                
    n_times = n_times - 1
    sleep(0.5) # This just adds a pause, of 0.5 seconds.

# This `print()` statement is outside of the `while` loop.
print('BOOM!')
````

The condition in the code cell above (`n_times > 0`) will be `True` until `n_times` equals 0 (duh!).

The code in the cell below is almost the same only *the condition in the `while` loop header has been changed*.
The condition is now `n_times > 5`. This means when `n_times` reaches 5, the condition will now be `False`.

An extra `print()` function has been added *after* the `while` loop, so it is easy to see *why* the `while` loop stopped running.

Look what happens to the countdown (look at the comment `LOOK HERE!` to see where the condition has been changed):

````python
n_times = 10

# This is the header of the `while` loop. 
while n_times > 5: # <--------------------------LOOK HERE!
    
    # This is the indented body of the `while` loop.
    print(f"{n_times}                                               `n_times > 0` = {n_times > 5}")
                                
    n_times = n_times - 1
    sleep(0.5) # This just adds a pause, of 0.5 seconds.

# These `print()` statements are outside of the `while` loop. (The extra `print()` function is the same
# as the one in the body of the `while` loop).
print(f"{n_times}                                               `n_times > 0` = {n_times > 5}") 
print('BOOM!')
````

Now, the `while` loop ends *as soon as the condition is `False`* (e.g. when `n_times` equals 5).
