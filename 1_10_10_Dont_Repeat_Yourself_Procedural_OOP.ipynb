# Don't Repeat Yourself

One very important principle of software development is [**D**on't **R**epeat **Y**ourself (DRY)](https://en.wikipedia.org/wiki/Don%27t_repeat_yourself). This principle states that:

- Programmers should avoid writing code with the same logic more than once. Instead they should write *functions* or *classes* which can be reused lots of times, without needing to be written again.

If we repeat code, we increase the chances of making mistakes when updating or rewriting it. Also, if other people work with you on your code, the more you repeat things, the more code there is. This makes the code harder to debug, test, and maintain.

**The benefit of following DRY is that it makes code easier to debug, test and maintain**. This is because, if each piece of logic is representing ONLY in one place, then if we need to change that logic, we only have to change it in one place. If it is spread over lots of places, then we have to change it in every place.

There are two major coding styles that can help us avoid repeating code, and so help us follow DRY:

**Procedural Programming** – Put code we will use lots of times into *functions*.

**Object-Oriented Programming (OOP)** – Put code we will use lots of times into classes and objects. Classes and objects are collections of attributes (data) and *methods*.

We have used the NumPy library a lot already. This library has examples in the Procedural Programming style and the Object Oriented Programming style.

The NumPy array is an *object* which means it is an instance (one example) of the array class:

```python
import numpy as np

prices = np.array([45, 65, 120])

type(prices)
```

`numpy.ndarray` is the *class*, our `price` array is one object which is an example of this class. The code inside NumPy for all of this is written in the object oriented style.

However, NumPy lets us (as programmers) use either style when we use NumPy. For example, we can calculate the `.mean()` price using the object oriented style - this is where we use NumPy *methods* which, as we know, are *functions attached to an **object***.

```python
prices.mean()
```

We can also use NumPy using the Procedural Programming style, but using its *functions*. So this is how we get in the mean in a more Procedural Programming coding style:

```python
np.mean(prices)
```

These both give us the same result, but the difference is in how we *get* the result:

- OOP style: `object.method()` (e.g.`prices.mean()`)

- Procedural Programming style: `function(object)` (e.g. `np.mean(prices)`)

## The Benefits of D.R.Y

Look at the following code:

```python
# Student data.
ages = np.array([21, 18, 18, 25, 26])
score_test_1 = np.array([77, 65, 99, 96, 88])
score_test_2 = np.array([68, 76, 92, 80, 81])

# Calculate the mean age.
mean_age = (ages[0] + ages[1] + ages[2] + ages[3] + ages[4])/len(ages)

# Calculate the average score on test 1.
mean_test_1 = (score_test_1[0] + score_test_1[1] + score_test_1[2] + score_test_1[3] + score_test_1[4])/len(ages)

# Calculate the average score on test 2.
mean_test_2 = (score_test_2[0] + score_test_2[1] + score_test_2[2] + score_test_2[3] + score_test_2[4])/len(ages)

print(f"Mean age = {mean_age} \nTest 1 Average = {mean_test_1} \nTest 2 Average = {mean_test_2}")
```

The problem with this code is that it doesn't follow DRY - every time the average of an array is calculated the logic for getting the mean is written again. **The chance of making a mistake gets higher the more we repeat ourselves like this**. For instance, maybe we will type an index wrong, maybe we will type the wrong operator by accident...

It is much safer and more efficient to use functions (Procedural Programming style) instead of repeating ourselves whenever we want to calculate the mean. We do that like this:

```python
# Student data.
ages = np.array([21, 18, 18, 25, 26])
score_test_1 = np.array([77, 65, 99, 96, 88])
score_test_2 = np.array([68, 76, 92, 80, 81])

# Calculate the mean age (using a NumPy function, in the Procedural Programming style).
mean_age = np.mean(ages)

# Calculate the average score on test 1 (using a NumPy function, in the Procedural Programming style).
mean_test_1 = np.mean(score_test_1)

# Calculate the average score on test 2 (using a NumPy function, in the Procedural Programming style).
mean_test_2 = np.mean(score_test_2)

print(f"Mean age = {mean_age} \nTest 1 Average = {mean_test_1} \nTest 2 Average = {mean_test_2}")
```

...or we can use NumPy methods (OOP style), like this:

```python
# Student data.
ages = np.array([21, 18, 18, 25, 26])
score_test_1 = np.array([77, 65, 99, 96, 88])
score_test_2 = np.array([68, 76, 92, 80, 81])

# Calculate the mean age (by NumPy method, in the OOP style).
mean_age = ages.mean()

# Calculate the average score on test 1 (by NumPy method, in the OOP style).
mean_test_1 = score_test_1.mean()

# Calculate the average score on test 2 (by NumPy method, in the OOP style).
mean_test_2 = score_test_2.mean()

print(f"Mean age = {mean_age} \nTest 1 Average = {mean_test_1} \nTest 2 Average = {mean_test_2}")
```

In either case:

- **we know these functions/methods work and have been tested, so reusing them instead of repeating the logic avoids errors.**

Here we are using functions and classes/methods written by the programmers who wrote NumPy. But the DRY principle also applies to *functions and classes that your write yourself*.
