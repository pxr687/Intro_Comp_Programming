## With statements

url = "https://www.google.com/"

from urllib.request import urlopen

urlopen(url)

type(urlopen(url))

[`with` automatically opens and closes a file]

with urlopen(url) as webpage:
    # Read the content and decode it from bytes to text
    webpage_as_text = webpage.read().decode('utf-8')
print(webpage_as_text[:2000])

from IPython.display import display, HTML

display(HTML(webpage_as_text))
