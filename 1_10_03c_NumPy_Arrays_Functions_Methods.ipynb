# NumPy Functions and Numpy Array Methods

Let's look a bit deeper at some of the things we can do with Numpy arrays, Numpy functions and Numpy array methods.

The main purpose of this page is to show you that *arrays are better than lists* for several reasons. You can think of Numpy arrays like the smarter, more attractive brother/sister of lists :) 

```python
# Load the Numpy package, and call it "np".
import numpy as np
```

"Repeat" means to do the same thing more than once.

The `np.repeat()` function will do this with strings or numbers. This is useful if we need to make lots of labels for something, or lots of numbers. Later, we will look at [simulation](https://en.wikipedia.org/wiki/Simulation), where this function can be very useful. 

```python
np.repeat('Hello', repeats = 10)
```

So, in the cell above:

- `np` is the name we are using to get the Numpy library, and the code in it

- `.repeat()` is the name of the Numpy function we want to use

- `'Hello'` is the string we want to repeat

- the `repeats = ` argument sets the number of repeats. 

If we want to repeat something 100 times then we use:

```python
np.repeat('Hello', repeats = 100)
```

We can also use lists as an input to this function (and lots of other Numpy functions too):

```python
# Make a list of names.
list_of_names = ['Pete', 'Paul', 'John']

list_of_names
```

```python
# Repeat the values in the list, 10 times.
np.repeat(list_of_names, 10)
```

Notice that this has not repeated the three names one after the other, like would happen if we use the `*` operator with the `list_of_names`:

```python
list_of_names * 10
```

Numpy is very flexible. If we want to repeat the names the way the `*` operator does with a list, then we can use the `np.tile()` function:

```python
# Repeat the values in the list, 10 times.
np.tile(list_of_names, reps = 10)
```

Look at the [documentation for `np.repeat()`](https://numpy.org/doc/2.3/reference/generated/numpy.repeat.html) and the [documentation for `np.tile()`](https://numpy.org/doc/2.3/reference/generated/numpy.tile.html).

We control the number of repeats using `repeats = ` for `np.repeat()` and `reps = ` for `np.tile()` even though each one does the same thing! Always check the documentation if you forget what the commands are for any function.

**So**, Numpy gives us two different ways to repeat the list, depending on which output we want for whatever task we are doing.

## NumPy functions for statistics/maths

Numpy also has a lot of functions for statistics and maths, which is why Numpy is a major Python data science library. Many of these statistics/maths functions can also be used as *methods*.

Imagine this is some data on student ages:

```python
array_of_ages = np.array([76, 23, 43])

array_of_ages
```

We can calculate the average age of these students using the `np.mean()` function:

```python
np.mean(array_of_ages)
```

We can get the total age (all of the ages added together) using the `np.sum()` function:

```python
np.sum(array_of_ages)
```

Remember that a *method* is a function which attached to an object.

We can use `.mean()` and `.sum()` as Numpy methods but putting them *after* the name of our array:

```python
# The `.mean()` method.
array_of_ages.mean()
```

```python
# The `.sum()` method.
array_of_ages.sum()
```

This is another advantage of Numpy arrays other Python lists - arrays have lots more mathematical/statistical methods, and is one of the reasons arrays are used more often than lists in data science.

We have already use `np.random.choice()`. We use this function by giving it a list or array as it's input. It then randomly selects elements from that list.

For instance, we can do a simulated coin flip with:

```python
np.random.choice(['HEADS', 'TAILS'])
```

Unless we tell it to use different probabilities, each element in the list has an equal chance of being selected. So in the cell above `"HEADS"` and `"TAILS"` have a 50% chance each of being chosen.

We can get random integers in the same way using `np.random.randint()`. We tell it what the lowest integer we want is using the `low = ` argument.

We tell it the highest integer we want using the `high = ` argument (note: we have to go one above the highest, so if the highest we want is 10, we should use `high = 11`):

```python
# Select a random integer betwee 0 and 1.
np.random.randint(low = 0, high = 2)
```

For most simulations, it is faster to use number data instead of strings.
