# Dictionaries

Dictionaries are another important type of Python collection.

```python
student_names = ['Limmy', 'Jen', 'Greg', 'Joseph']

student_grades = [90, 87, 66, 78]
```

```python
student_names[0]
```

```python
student_grades[0]
```

```python
student_dict = {'Name' : student_names,
                'Grade': student_grades}

student_dict
```

```python
student_dict['Name']
```

```python
student_dict['Grade']
```

```python
student_dict.keys()
```

```python
student_dict.values()
```

```python
student_dict['Name'][0]
```

```python
print("Student 1 = ", student_dict['Name'][0], ", their grade is", student_dict['Grade'][0], ".")
```

[Lots of API calls return data that is best used as a dictionary]

```python
from urllib.request import urlopen
import json

flip_a_coin_API = "https://flipacoinfree.com/api/flip"

with urlopen(flip_a_coin_API) as response:
    coin_flip_from_API = json.loads(response.read().decode("utf-8"))

coin_flip_from_API
```

```python
coin_flip_from_API['data']
```

```python
coin_flip_from_API['data']['result']
```
