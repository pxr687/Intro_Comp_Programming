## Scientific notation - another way of writing float values

`float` values in scientific notation are in *exponential* format. We can make a literal in exponential format like this:

````python
# Two thousand one hundred.
2.1e3
````

As mentioned above, this exponential format is called *scientific notation*.

`2.1e3` is written in scientific notation, and it is the same value as:

````python
2100.0
````

You can use the `format()` function to make Python show you the number NOT in scientific notation. This is shown in the cell below:

````python
# Take a number in scientific notation, and show it NOT in scientific notation.
format(2.1e3, "f")
````

The `"f"` in the `format()` function tells it that we want [*fixed-point notation*](https://docs.python.org/3/library/string.html). You can think of this as just meaning "NOT scientific notation"! **Note**: this will only show you 6 decimal places, unless you change other settings...(e.g. you can change `"f"` to `".24f"` to show 24 decimal places...).

````python
# Make Python show us 24 decimal places.
format(2.1e3, ".24f")
````

Also note that `format()` returns your number as text/string data, not as `float` data - you will get errors if you ignore this!

Ok, back to scientific notation. 

You can read `2.1e3` as $2.1 * 10^3$.  The `e3` means "times 10 to the power of 3".

So we say `2.1e3` as "2.1 times 10 to the power of 3".

We can also use negative powers to represent small numbers:

````python
1e-3
````

Read `1e-3` as $1 x 10^-3$, or "1, but move the decimal point 3 spaces to the
left", or:

````python
1 * 10 ** -3
````

## The limits of float values

> Limit, *noun*, ["the greatest or smallest amount of something that is allowed."](https://www.oxfordlearnersdictionaries.com/definition/english/limit_1)

Float values are very flexible, but they do have limits.

1. A `float` can represent very large and very small numbers. There
   are limits, but you will rarely encounter them.
2. A `float` only represents 15 or 16 significant digits for any number; the
   remaining digits are lost. This limited precision is enough for most uses.
3. After putting together `float` values with mathematical operators, the last few digits may be
   incorrect. Small rounding errors are often confusing when first
   encountered.

The first limit can be seen in two ways. If the result of a calculation is
a very large number, then it is represented as infinite. If the result is a
very small number, then it is represented as zero.

````python
# Very big, but can still be represented by the computer.
2e306 * 10
````

````python
# Too big! Comes back as `inf` for "infinite"
2e306 * 100
````

````python
# Very small, but can still be represented by the computer.
2e-322 / 10
````

````python
# Too small! Comes back as 0.
2e-322 / 100
````

The second limit can be seen by an expression that involves numbers with
more than 15 significant digits. These extra digits are thrown away before any
arithmetic is carried out.

````python
# Our very small digits are NOT lost.
0.666666 - 0.666666123456789
````

````python
# Our very small digits ARE lost :(
0.6666666666666666 - 0.6666666666666666123456789 # The digits on the end of the second number are thrown away.
````

The third limit can be seen when calculating the difference between two
expressions that should be equivalent. For example, the expression `2 ** 0.5`
computes the square root of 2, but squaring this value does not exactly
make 2. 

Here we `import` the function `math.sqrt()`, which returns the square root of numbers it is given as arguments.

````python
import math
math.sqrt(2)
````

Remember that the square root of 2 is the number that we can multiply by itself to get 2 e.g. 

$\text{square root of 2} * \text{square root of 2} = 2$

````python
# The square root of 2∗square root of 2
1.4142135623730951 * 1.4142135623730951
````

````python
# The square root of 2∗square root of 2
math.sqrt(2) * math.sqrt(2)
````

Let's subtract `2` to get that tiny fractional component by itself:

````python
# The the tiny number by itself.
math.sqrt(2) * math.sqrt(2) - 2
````

The final result above is `0.0000000000000004440892098500626`, a number that
is very close to zero. The correct answer to this arithmetic expression is 0,
but a small error in the final significant digit appears very different in
scientific notation.

This behavior appears in almost all programming languages
because it is the result of the standard way that arithmetic is done by
computers. This is called **floating point error** e.g. very small errors which
happen because of floating point numbers.

Although `float` values are not always exact, they are predictable and work
in the same way across all different kinds of computers and programming
languages.

In summary: 

- doing calculations with very big or very small numbers can lead to floating point errors.

## Before you do the exercise

Make sure you understand this (it can be confusing, so read it carefully):

Think about `4.2e+05`. This base number (`4.2`) is positive, and it has a plus sign `+` in it's scientific notation (e.g. `e+05`). This is a *positive number far from 0*:

````python
4.2e+05
````

However, the number `-4.2e+05` number is *negative*, it has a minus sign `-` at it's start. But it has a plus sign `+` in it's scientific notation (e.g. `e+05`). This is a *negative number far from 0*:

````python
format(-4.2e+05, "f")
````

The number `4.2e-05` is positive, it does NOT have a minus sign at it's start. BUT it does have a minus sign in its scientific notation (e.g. `e-05`). So, this is a *positive number close to 0*:

````python
format(4.2e-05, ".24f")
````

In contrast, the number `-4.2e-05` has a minus sign `-` at it's start, AND in it's scientific notation (e.g. `e-05`). This is a *negative number close to 0*:

````python
format(-4.2e-05, ".24f")
````

So:

| Base Number Sign | `e` Sign | Meaning | Example |
|---|---|---|---|
| Positive (`+`) | Positive (`+`) | **Positive number far from 0** | `4.2e+05` = `420000` |
| Negative (`-`) | Positive (`+`) | **Negative number far from 0** | `-4.2e+05` = `-420000` |
| Positive (`+`) | Negative (`-`) | **Positive number close to 0** | `4.2e-05` = `0.000042` |
| Negative (`-`) | Negative (`-`) | **Negative number close to 0** | `-4.2e-05` = `-0.000042` |

If `e` is POSITIVE (+)  →  move decimal RIGHT  →  number is farther from 0

If `e` is NEGATIVE (-)  →  move decimal LEFT   →  number is closer to 0

The sign of the base number determines which side of 0 it is on, look at the number line below. The `X`'s show the location of different numbers, along with their scientific notation:
