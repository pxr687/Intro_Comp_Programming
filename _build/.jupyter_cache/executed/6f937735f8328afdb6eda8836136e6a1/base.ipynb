```python
N = 10000

head_counts = np.empty(N, dtype='object')

for i in np.arange(N):
    tosses = np.random.choice(["HEADS", "TAILS"], size=100)
    n_heads = np.sum(tosses == "HEADS")
    head_counts[i] = n_heads

head_counts
```

```python
# Import `matplotlib`.
import matplotlib.pyplot as plt
```

```python
# Make a histogram, with some custom options.
plt.hist(head_counts, bins=np.arange(30.5, 69.6, 1))
plt.xlabel('Number of Heads')
plt.ylabel('Count');
```
