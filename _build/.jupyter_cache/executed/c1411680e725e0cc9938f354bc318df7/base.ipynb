```python
# An array of numbers.
my_array = np.array([1, 5, 19, 2000])

# Make a Boolean array - `True` if the number in `my_array` is > 10, `False` otherwise.
my_boolean_array = my_array > 12

# Use the Boolean array to index `my_array`.
my_array[my_boolean_array]
```

```python
False == 0
```

```python
ages_of_students = np.array([18, 24, 22, 19, 28, 19])

ages_of_students
```

```python
# A Boolean array.
ages_of_students > 20
```

```python
# The number of students older than 20.
np.sum(ages_of_students > 20)
```

```python
# The number of students older than 20 (another way).
np.count_nonzero(ages_of_students > 20)
```
