```python
import numpy as np

for name in ["Yan", "Emily", "Riley", "Pooh", "Ryan"]:
    print(name)
```

```python
random_name = np.random.choice(["Emily", "Riley", "Pooh"])
random_name
```

```python
for name in ["Yan", "Emily", "Riley", "Pooh", "Ryan"]:
    if name == random_name:
        break
    print(name)

print("End of loop!")
```

```python
for name in ["Yan", "Emily", "Riley", "Pooh", "Ryan"]:
    if name == random_name:
        continue
    print(name)

print("End of loop!")
```
