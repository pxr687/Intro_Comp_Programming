```python
# DO NOT CHANGE THIS CODE, JUST RUN IT. The comments
# explain what is happening.

# Import the `requests` and `numpy` libraries.
import requests
import numpy as np

# A string variable with the URL of the API we will use.
url = "https://random-word-api.herokuapp.com/word?number=1"

# This code just gets our word, and makes sure it is just one string.
username = requests.get(url)
username = username.json()[0]
# Add some randomness to the word (users often break rules)!
if np.random.choice([True, False]):
    username = np.random.choice([username.capitalize(), username.upper(),
                                 username+str(np.random.randint(0, 2000)),
                                 username[:3], username[:4], username[:5]])

# Show the word.
print(username)
```
