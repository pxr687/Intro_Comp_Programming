```python
import numpy as np
from time import sleep # We will use this later on.

n_times = 10

for i in np.arange(n_times):
    
    # We know this `print()` statement will repeat 10 times.
    print("Count me!")
```

```python
n_times = 10

# This is the header of the `while` loop. 
while n_times > 0:
    
    # This is the indented body of the `while` loop.
    print(f"Countdown: {n_times}                                               `n_times > 0` = {n_times > 0}")
                                
    n_times = n_times - 1
    sleep(0.5) # This just adds a pause, of 0.5 seconds.

# This `print()` statement is outside of the `while` loop.
print('BOOM!')
```

```python
n_times = 10

# This is the header of the `while` loop. 
while n_times > 5: # <--------------------------LOOK HERE!
    
    # This is the indented body of the `while` loop.
    print(f"Countdown: {n_times}                                               `n_times > 0` = {n_times > 5}")
                                
    n_times = n_times - 1
    sleep(0.5) # This just adds a pause, of 0.5 seconds.

# These `print()` statements are outside of the `while` loop. (The extra `print()` function is the same
# as the one in the body of the `while` loop).
print(f"Condition is now false at: {n_times}                               `n_times > 0` = {n_times > 5}") 
print('BOOM!')
```
