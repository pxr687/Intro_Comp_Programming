```python
# A one-line string.
print("This is a string on one line.")
```

```python
# A multi-line string.
print( """
This is a string...

...which goes over lots of...

...lines, I can go over...

...as many lines as I want.

To make a multi-line string...

...you use THREE quotation marks...

...to start and end the string.
""")
```

```python
# You can also store multi-line strings in variables.
my_multi_string = """
This is a string...

...which goes over lots of...

...lines, I can go over...

...as many lines as I want.

To make a multi-line string...

...you use THREE quotation marks...

...to start and end the string.
"""

# `print()` the multi-line string variable.
print(my_multi_string)
```

```python
# Make the `my_string` variable.
my_string = "Poto1"

# Show the value of `my_string`.
my_string
```

```python
# Use the `.replace()` string method. 
# Here I replace the character 'o' with the character 'e'.
my_string.replace('o', 'e')
```

```python
# We can use replace lots of times with the same string.
# Here I replace the character 'o' with the character 'e' AND `1` with `r`.
my_string.replace('o', 'e').replace('1', 'r')
```

```python
# We can use `.replace()` lots of times AND save the result in a variable.
my_fixed_string = my_string.replace('o', 'e').replace('1', 'r')

# Show the value of `my_fixed_string`.
my_fixed_string
```
