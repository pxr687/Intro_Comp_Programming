```python
# Import `pandas`.
import pandas as pd

# This reads in the file and stores it as a variable.
df = pd.read_csv('data/airline_passengers.csv')

df
```

```python
# Import `matplotlib`, rename it as `plt`.
import matplotlib.pyplot as plt

plt.hist(df['Thousands of Passengers'])
```

```python
plt.hist(df['Thousands of Passengers'])

# Add some axis labels.
plt.xlabel('Thousands of Passengers')
plt.ylabel('Count')

# Add a title.
plt.title('Thousands of Passengers')

# This stops the junk at the top of the plot.
plt.show()
```

```python
# Create a figure and axes. These are instances of classes.
# This line of conde makes the instances.
fig, ax = plt.subplots()

# Plot the histogram (notice that now we are using *methods* of the axes objects, instead of functions).
ax.hist(df['Thousands of Passengers'])

# Add axis labels (notice that now we are using *methods* of the axes objects, instead of functions).
ax.set_xlabel('Thousands of Passengers')
ax.set_ylabel('Count')

# Add a title (notice that now we are using *methods* of the axes objects, instead of functions).
ax.set_title('Thousands of Passengers')

# Display the plot.
plt.show()
```

```python
# Make the scatterplot.
plt.scatter(x = df['Month'],
            y = df['Thousands of Passengers'])

# Add axis labels.
plt.xlabel('Month')
plt.ylabel('Thousands of Passengers')

# Show the plot.
plt.show()
```
