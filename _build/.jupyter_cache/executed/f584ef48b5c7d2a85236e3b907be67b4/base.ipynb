```python
# Classify without the unnecessary first assignment.
x = np.random.randint(low = -10, high = 11)

if x > 0:
    result = 'Positive'
    
else:
    result = 'Not positive'

print(x)
print(result)
```

```python
x = np.random.randint(low = -10, high = 11)

if x > 0:
    result = 'Positive'
    
elif x == 0:  # elif short for "else if"
    result = 'Zero'
    
else:
    result = 'Negative'
    
print(x)
print(result)
```
