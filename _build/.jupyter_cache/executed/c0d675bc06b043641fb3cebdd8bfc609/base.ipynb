```python
# Get the text for Pride and Prejudice.

# Import a library to get data from the internet.
import urllib.request

# Make a string variable which is a website address.
book_url = 'http://www.gutenberg.org/ebooks/42671.txt.utf-8'

# Get the book from the website.
book_text = urllib.request.urlopen(book_url).read().decode()

# Show the first 5000 lines (do not worry about understanding this code for now).
print(book_text[:5000])
```
