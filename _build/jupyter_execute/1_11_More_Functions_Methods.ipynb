## More on Functions

```python
import numpy as np
```

```python
ages_of_class = [20, 22, 22, 23, 21, 17, 21, 15, 21, 16, 24, 24, 19, 22, 21]

ages_of_class
```

```python
import matplotlib.pyplot as plt

plt.hist(ages_of_class)
plt.xlabel('Age (Years)')
plt.ylabel('Number of Students');
```

```python
np.mean(ages_of_class)
```

```python
print("The average age of students in the class is", np.mean(ages_of_class), "years old.")
```

Convert age in years to months?

```python
ages_of_class * 12
```

```python
len(ages_of_class * 12)
```

```python
180/12
```

```python
180/12 == len(ages_of_class)
```

```python
ages_of_class_array = np.array(ages_of_class)
```

```python
ages_of_class_array
```

```python
ages_of_class_array * 12
```

```python
ages_in_months = ages_of_class_array * 12
```

```python
plt.hist(ages_in_months)
plt.xlabel('Age (Months) - Approximate')
plt.ylabel('Number of Students');
```

```python
print("The average age of students in the class is", np.mean(ages_in_months), "months old.")
```

## Methods

Methods are like a different flavour of function...

```python
ages_in_months.mean()
```

```python
np.mean(ages_in_months)
```

```python
type(ages_in_months)
```

`dir(ages_in_months)` is messy but it will show you all the methods available on Numpy array...

```python
ages_in_months.max()
```

```python
np.max(ages_in_months)
```

```python
ages_in_months.min()
```

```python
np.min(ages_in_months)
```
