When you have a string, you can make new strings from it using string methods. String methods are *functions* that *work on strings*. These methods are used by putting a dot after the string, then calling the function.

For example, the following method generates an uppercase (all capitals) version of a string.

````python
# Use the `.upper()` string method with the string `loud`.
"loud".upper()
````

````python
# Use the `.upper()` string method with the string `shouting`.
"shouting".upper()
````

We already saw the next string method on an earlier page. It is probably the most important string method - it is called `.replace()`. What it does is *replace* a substring within the string. The `.replace()` method takes two arguments, the text to replace and the text to replace it with.

````python
# Use the `.replace()` string method with the string `"sunny"`.
'sunny'.replace('s', 'f')
````

````python
# Use the `.replace()` string method with the string `"drink"`.
'drink'.replace('dr', 'th')
````

String methods can also be used with variables, if the variable is a string. 

So, for instance, the following code cell makes the word "money" starting from "many" by first making "my", and then doing another replacement after that.

````python
# Our first word, "many".
first_word = "many"

# Use `.replace()` to make our second word ("my").
second_word = first_word.replace('an', '')

# Use `.replace()` to make our third word ("money")
third_word  = second_word.replace('y', 'oney')
third_word
````

Let's break that down to make it clearer what is happening:

````python
# The string we start with.
first_word
````

````python
# Use `.replace()` to make our second word ("my").
second_word = first_word.replace('an', '')
second_word
````

````python
# Replace "in" with "de"
third_word  = second_word.replace('y', 'oney')
third_word
````

Notice that the line `second_word = first_word.replace('an', '')` does **NOT** change the string `first_word`,
which is still "many". 

````python
first_word
````

## Practical Uses of String Methods

Strings methods have many uses. One common use is to make sure that your text data is *consistent*.

E.g. maybe you want all of your filenames to be in the same format, with no blank spaces:

````python
# Original filename (has spaces in).
filename = "My Summer Photo.JPG"
````

````python
# Using string methods to "clean" the filename, so that it does not have any spaces in.
filename_clean = filename.replace(' ', '_')

filename_clean
````

For another example, think about if a user entered this email address, into an online form:

````python
email = "  Jack.Jones@CrazyMail.COM  "

email
````

Maybe you want to write a program which will send marketing emails for your company. The email address, the way it is now, will need to be changed so that a program can use it properly. 

It has space, and capital letters in random places. We can make it consistent using string methods:

````python
# Clean the email address.
email_clean = email.strip()
email_clean = email_clean.lower()
email_clean = email_clean.replace(" ", "")
email_clean
````

The cell above *re-assigns* the `email_clean` variable each time a new string method is used. We can do this more efficiently by *chaining* string methods together...

## Chaining String Methods 

You can also *chain* string methods together, one after the other. Think about this string:

````python
"this sentence looks bad. there should be more capital letters!"
````

If we want to alter it to `"This sentence no longer looks bad. There are enough capital letters!"` we can chain multiple string methods together:

````python
# Chaining string methods together...
"this sentence looks bad. there should be more capital letters!".capitalize().replace('. t', '. T').replace('looks', 'no longer looks').replace('should be more', 'are enough')
````

We can make that same operation easier (for a human) to read, but placing it over more than one line:

````python
"this sentence looks bad. there should be more capital letters!".capitalize().replace('. t',
                                                                                      '. T').replace('looks',
                                                                                                     'no longer looks').replace('should be more',
                                                                                                                                'are enough')
````
