## Boolean Arrays and Boolean Indexing

Another advantage of Numpy arrays is that we can use them in expressions with comparison operators (`<`, `>`, `!=` etc).

Let's say we want to ask Python "which elements in this collection are bigger than 10?".

If we try the following code with a list, we get a nasty error message:

```python
my_list = [1, 5, 19, 2000]
my_list > 10
```

On the other hand, this works if our collection is an array:

```python
import numpy as np
my_array = np.array([1, 5, 19, 2000])
my_array > 10
```

The output of the cell above is a *Boolean array* - this means it is an array which has only Boolean values. For this array we get a `True` value at the places where the original `my_array` has a value larger than 10, and a `False` at places where the value was less than 10. This is show more clearly below:

| Index | 0 | 1 | 2 | 3 |
|---|---:|---:|---:|---:|
| Value | 1 | 5 | 19 | 2000 |
| Comparison | `1 > 10` | `5 > 10` | `19 > 10` | `2000 > 10` |
| Result | `False` | `False` | `True` | `True` |

The first row of the table shows the values in the original `my_array`.

The middle row shows the comparison using `> 10`.

The last row shows the result in the Boolean array.

So, arrays are better than lists here because we can more easily ask questions about their values using comparison operators.


## Filtering NumPy Arrays with Boolean Arrays

One way that Boolean arrays are super useful is that they let us get values out of others arrays, where those values pass a test.

For instance, we can use the Boolean array we just saw to **get only the values larger than 10** from `my_array`.

Here is `my_array` again:

```python
my_array
```

We can make the Boolean array using a simple comparison expression:

```python
my_boolean_array = my_array > 10

my_boolean_array
```

We can then do something callen *Boolean indexing* - this is where we use the Boolean array to index another array, like this:

```python
my_array[my_boolean_array]
```

You can think of the `True` and `False` values in the Boolean array as "switches". 

If the value in a Boolean array is `False`, then when we use it for Boolean indexing, the same value in the array we are indexing gets "switched off". This means it does NOT end up in the array we get as a result of Boolean indexing.

Conversely, ff the value in a Boolean array is `True`, then when we use it for Boolean indexing, the same value in the array we are indexing gets "switched ON". This means it DOES end up in the array we get as a result of Boolean indexing.

So, we can do this all in one cell with:

```python
# An array of numbers.
my_array = np.array([1, 5, 19, 2000])

# Make a Boolean array - `True` if the number in `my_array` is > 10, `False` otherwise.
my_boolean_array = my_array > 12

# Use the Boolean array to index `my_array`.
my_array[my_boolean_array]
```

The table shows what has happened here:

| Index | 0 | 1 | 2 | 3 |
|---|---:|---:|---:|---:|
| `my_array` | 1 | 5 | 19 | 2000 |
| `my_boolean_array = my_array > 12` | `False` | `False` | `True` | `True` |
| `my_array[my_boolean_array]` | — | — | 19 | 2000 |

The `False` values act as "switches" which switch off the values in `my_array` which are less than 10.

So the last array we get only has the values from `my_array` which are bigger than 10.

**This works exactly the same way with string/text data**.

For example, an array of the names of the biggest cities in Thailand:

```python
cities = np.array(["Bangkok", "Chiang Mai", "Nakhon Ratchasima", "Khon Kaen", "Hat Yai"])
cities
```

We can make Boolean arrays using any of the comparison operators:

```python
is_it_BKK = cities == "Bangkok"

is_it_BKK
```

...then use the Boolean array to index the original array, to filter the `cities`. Remember, that the `True` values act as ON switches, and the `False` values act as OFF switches:

```python
cities[is_it_BKK]
```

We can use a different comparison operator, to do a different filtering operation - for instance, to get all of the `cities` which are NOT Bangkok:

```python
is_it_NOT_BKK = cities != "Bangkok"

is_it_NOT_BKK
```

...then, use the new `is_it_NOT_BKK` to get all of the other cities:

```python
cities[is_it_NOT_BKK]
```

These Boolean arrays can be used to filter other arrays **which come from the same data**.

This is **important**: the data must *line up* between arrays.

So here is our `cities` array:

```python
cities
```

Here is another array representing the `population` of each city:

```python
population = np.array([10539000, 1198000, 470000, 418000, 407000])

population
```

The data at each index location is for one city:

```python
# Bangkok data.
print(cities[0], population[0])
```

```python
# Chiang Mai data.
print(cities[1], population[1])
```

So we can use Boolean indexing *on the `cities` array* to get the `population` of any city we want:

```python
is_it_BKK
```

```python
# Get the `population` of Bangkok.
population[is_it_BKK]
```

This is a very important concept:

 > If your arrays get "out of line", then **your data will become meaningless**.
 
As your write more complex data analysis programs, this can easily happen, so always check back to your *original data*.

## Using `.sum()` will count the number of `True` values in an array

It is important to know that Python will treat `True` values as being equal to 1:

```python
True == 1
```

And `False` values as being equal to 0:

```python
False == 0
```

This means that when we use a comparison operator, and it gives us a Boolean array, we can easily count the number of times something is `True` just by using addition.

Imagine we have some data on the ages of students in a class:

```python
ages_of_students = np.array([18, 24, 22, 19, 28, 19])

ages_of_students
```

Imagine we want to contact students over the age of 20, to get them to apply for an internship which only people older than 20 can apply for.

So, we want to ask Python *"how many students are older than 20?"* so we know how many people we need to contact?

We can do this easily with a comparison operator and `np.sum()`:

```python
# A Boolean array.
ages_of_students > 20
```

We can give this Boolean array straight to the `np.sum()` function:

```python
# The number of students older than 20.
np.sum(ages_of_students > 20)
```

We see the number 3 because Python has counted the `True` values - this means 3 students are older than 20.

We can also do this with the `np.count_nonzero()` function:

```python
# The number of students older than 20 (another way).
np.count_nonzero(ages_of_students > 20)
```
