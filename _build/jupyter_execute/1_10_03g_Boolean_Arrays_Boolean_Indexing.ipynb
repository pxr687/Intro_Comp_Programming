## Boolean Arrays and Boolean Indexing

Another advantage of Numpy arrays is that we can use them in expressions with comparison operators.

Let's say we want to ask Python "which elements in this collection are bigger than 10?".

If we try the following code with a list, we get a nasty error message:

```python
my_list = [1, 5, 19, 2000]
my_list > 10
```

On the other hands, this works if our collection is an array:

```python
my_array = np.array([1, 5, 19, 2000])
my_array > 10
```

The output of the cell above is a *Boolean array* - this means it is an array which has only Boolean values. For this array we get a `True` value at the places where the original `my_array` has a value larger than 10, and a `False` at places where the value was less than 10. This is show more clearly below:

| Index | 0 | 1 | 2 | 3 |
|---|---:|---:|---:|---:|
| Value | 1 | 5 | 19 | 2000 |
| Comparison | `1 > 10` | `5 > 10` | `19 > 10` | `2000 > 10` |
| Result | `False` | `False` | `True` | `True` |

The first row of the table shows the values in the original `my_array`.

The middle row shows the comparison using `> 10`.

The last row shows the result in the Boolean array.

So, arrays are better than lists here because we can more easily ask questions about their values using comparison operators.


## Filtering NumPy Arrays with Boolean Arrays

One way that Boolean arrays are super useful is that they let us get values out of others arrays, where those values pass a test.

For instance, we can use the Boolean array we just saw to **get only the values larger than 10** from `my_array`.

Here is `my_array` again:

```python
my_array
```

We can make the Boolean array using a simple comparison expression:

```python
my_boolean_array = my_array > 10

my_boolean_array
```

We can then do something callen *Boolean indexing* - this is where we use the Boolean array to index another array, like this:

```python
my_array[my_boolean_array]
```

You can think of the `True` and `False` values in the Boolean array as "switches". 

If the value in a Boolean array is `False`, then when we use it for Boolean indexing, the same value in the array we are indexing gets "switched off". This means it does NOT end up in the array we get as a result of Boolean indexing.

Conversely, ff the value in a Boolean array is `True`, then when we use it for Boolean indexing, the same value in the array we are indexing gets "switched ON". This means it DOES end up in the array we get as a result of Boolean indexing.

So, we can do this all in one cell with:

```python
# An array of numbers.
my_array = np.array([1, 5, 19, 2000])

# Make a Boolean array - `True` if the number in `my_array` is > 10, `False` otherwise.
my_boolean_array = my_array > 12

# Use the Boolean array to index `my_array`.
my_array[my_boolean_array]
```

The table shows what has happened here:

| Index | 0 | 1 | 2 | 3 |
|---|---:|---:|---:|---:|
| `my_array` | 1 | 5 | 19 | 2000 |
| `my_boolean_array = my_array > 12` | `False` | `False` | `True` | `True` |
| `my_array[my_boolean_array]` | — | — | 19 | 2000 |

The `False` values act as "switches" which switch off the values in `my_array` which are less than 10.

So the last array we get only has the values from `my_array` which are bigger than 10.
