# How this class is graded

The assessments for this class are spread out over the semester:

| Week | Assessment | % of Grade |
| :--: | :---------: | :----: |
| 2 | In-class Assignment | 5% |
| ↓ | | |
| 4 | Test #1 | 15% |
| ↓ | | |
| 7 | In-class Assignment | 5% |
| ↓ | | |
| 8 | **Midterm Exam** | **25%** |
| ↓ | | |
| 11 | In-class Assignment | 5% |
| ↓ | | |
| 12 | Test #2 | 20% |
| ↓ | | |
| 15 | **Final Exam** | **25%** |

Your final grade is made up from all of these assessments.
