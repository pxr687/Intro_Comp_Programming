Computers are very good calculators. However, there are important details  that every programmer working with numbers needs to know. Python (and most other programming languages) uses two *different types of numbers*. We have seen both of them on the [Data Types](1_08_00_Data_Types) page:

* Integers are called `int` values in the Python language. They can only
  represent whole numbers (negative, zero, or positive) that do not have a decimal place. E.g. `1`, `100`, `0` and `-1` are all integers.
  
* Numbers with decimal places are called `float` values (or *floating point values*) in the
  Python language. They can represent whole numbers or numbers with decimal placees. E.g. `10.6`, `1000.0292`, `-17.5` and $\frac{90}{100}$ would all be `float` values in Python, because they all have decimal places. *Note*: decimal places are also called a *fractional component*.

The type of a number can be seen from the way it is displayed: `int` values
have no decimal point and `float` values always have a decimal point.

Here are some `int` values:

````python
2
````

````python
1 + 3
````

````python
-1234567890000000000
````

Here are some float values:

````python
1.2
````

````python
1.5 + 2
````

````python
3 / 1
````

````python
-12345678900000000000.0
````

When a `float` value is combined with an `int` value using some mathematical
operator, then the result is always a `float` value:

````python
7.5 + 4
````

````python
# A number with a decimal of 0 is still a `float`!
8.0 + 2
````

````python
70 * 102.3
````

In most cases, two integers put together will make another integer, but any number (`int` or `float`)
divided by another will be a `float` value: 

````python
# Add two integers.
6 + 4
````

````python
# Divide two integers (this returns a `float`).
100/10
````

## Scientific notation - another way of writing float values

You can also write float values in *exponential* format, like this:

````python
# Two thousand one hundred.
2.1e3
````

This is the same value as:

````python
2100.0
````

Read `2.1e3` as $2.1 x 10^3$.  The `e3` means "times 10 to the power of 3".

We can also use negative powers to represent small numbers:

````python
1e-3
````

Read `1e-3` as $1 x 10^-3$, or "1 then moving the decimal point 3 spaces to the
left", or:

````python
1 * 10 ** -3
````

## The limits of float values

Float values are very flexible, but they do have limits.

1. A `float` can represent very large and very small numbers. There
   are limits, but you will rarely encounter them.
2. A `float` only represents 15 or 16 significant digits for any number; the
   remaining digits are lost. This limited precision is enough for most uses.
3. After putting together `float` values with mathematical operators, the last few digits may be
   incorrect. Small rounding errors are often confusing when first
   encountered.

The first limit can be seen in two ways. If the result of a calculation is
a very large number, then it is represented as infinite. If the result is a
very small number, then it is represented as zero.

````python
# Very big, but can still be represented by the computer.
2e306 * 10
````

````python
# Too big! Comes back as `inf` for "infinite"
2e306 * 100
````

````python
# Very small, but can still be represented by the computer.
2e-322 / 10
````

````python
# Too small! Comes back as 0.
2e-322 / 100
````

The second limit can be observed by an expression that involves numbers with
more than 15 significant digits. These extra digits are discarded before any
arithmetic is carried out.

````python
# Our very small digits are NOT lost.
0.666666 - 0.666666123456789
````

````python
# Our very small digits ARE lost :(
0.6666666666666666 - 0.6666666666666666123456789
````

The third limit can be observed when taking the difference between two
expressions that should be equivalent. For example, the expression `2 ** 0.5`
computes the square root of 2, but squaring this value does not exactly
recover 2. 

Here we `import` the function `math.sqrt()`, which returns the square root of numbers it is given as arguments.

````python
import math
math.sqrt(2)
````

Remember that the square root of 2 is the number that we can multiply by itself to get 2 e.g. 

$\text{square root of 2} * \text{square root of 2} = 2$

````python
# The square root of 2∗square root of 2
1.4142135623730951 * 1.4142135623730951
````

````python
# The square root of 2∗square root of 2
math.sqrt(2) * math.sqrt(2)
````

Let's subtract `2` to get that tiny fractional component by itself:

````python
# The the tiny number by itself.
math.sqrt(2) * math.sqrt(2) - 2
````

The final result above is `0.0000000000000004440892098500626`, a number that
is very close to zero. The correct answer to this arithmetic expression is 0,
but a small error in the final significant digit appears very different in
scientific notation.

This behavior appears in almost all programming languages
because it is the result of the standard way that arithmetic is done by
computers. This is called **floating point error** e.g. very small errors which
happen because of floating point numbers.

Although `float` values are not always exact, they are predictable and work
in the same way across all different kinds of computers and programming
languages.

In summary: 

- doing calculations with very big or very small numbers can lead to floating point errors.
