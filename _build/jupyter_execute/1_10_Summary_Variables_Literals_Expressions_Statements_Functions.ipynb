# Summary: what we have learned so far.

- A **literal** is...

- An **expression** is...

- A **variable** is...

- A **statement** is...

- A **function is**....

Let's summarise what we have learned so far. We will add each Python *element* (a thing in Python grammar) to a table. On earlier pages, we talked about variables and expressions:

| Python Element | What it is/does | Example 1 |  Example 2 |
| --- | --- | --- |  --- |  
| Variable | Stores data in our computer's memory | `rent = 1100` | `favourite_fruit = "mango"`
| Expression | Piece of Python code which returns a value | `100 + 100` |  `(900*5)/6`

A *statement* - like a variable assignment statement - is [asking Python to *do* something](https://www.w3schools.com/python/python_statements.asp) (e.g. "make a variable!").

An *expression* is asking Python to *show us something* (e.g. "what is the value of `100 * 2`?", or "what is the value of the variable `number_3`?"
