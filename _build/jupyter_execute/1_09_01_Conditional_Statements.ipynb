import numpy as np
from IPython.display import display, HTML

roll = np.random.randint(1, 7)

student_1_name = "Dave"

student_2_name = "Krystal"

student_2_money = 100

student_1_bet = 3

student_2_bet = 5

# Random dice roll, then win/lose money
roll = np.random.randint(1, 7)

roll_to_index_list = roll - 1

dice = ["⚀", "⚁", "⚂", "⚃", "⚄", "⚅"][roll_to_index_list]

jupyprint(f'<h1>{dice}</h1><p>The roll was <b style="color:red;">{roll}</b></p>')

student_1_diff = abs(student_1_bet - roll)

student_2_diff = abs(student_2_bet - roll)

jupyprint(f'<b>{student_1_name} bet {student_1_bet}.</b>')
jupyprint(f'<b>{student_2_name} bet {student_2_bet}.</b>')

if student_1_diff == student_2_diff:
    jupyprint(f'<h1>Draw!</h1>')
    
elif student_1_diff > student_2_diff:
    jupyprint(f'<h2>{student_2_name} wins $20!</h2>')
    student_2_money = student_2_money + 20
    jupyprint(f'<h2>{student_1_name} loses $20!</h2>')
    student_1_money = student_1_money - 20
    
elif student_1_diff < student_2_diff:
    jupyprint(f'<h2>{student_1_name} wins $20!</h2>')
    student_1_money = student_1_money + 20
    jupyprint(f'<h2>{student_2_name} loses $20!</h2>')
    student_2_money = student_2_money - 20
    
jupyprint("_________________________________________________________")
display(HTML(f'<b>{student_1_name} has ${student_1_money} left.</b>'))
display(HTML(f'<b>{student_2_name} has ${student_2_money} left.</b>'))

if (student_1_money <= 0) or (student_2_money <= 0):
    jupyprint("# GAME OVER!")
