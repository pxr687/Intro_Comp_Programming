How does Jupyter Notebook work? It has two parts:

* The web client (you are looking at this now, if you are running this
  notebook in Jupyter);
* The kernel (this runs the Python code, and makes the results).

Look at this code below:

````python
# Print some text.
print('Hello world!')
````

I type the code into the web client. I press `Shift + Enter`, or click the `Run` button and this sends a *message* to the *kernel*.

The message says something like:

> Here is some Python code.  Run this code, and show me the results.

The kernel runs the code in Python, and then sends back the results, in this case, as text `Hello world!`. The web client shows the results.

![](https://raw.githubusercontent.com/pxr687/Intro_Comp_Programming/refs/heads/main/images/how_notebook_works.png)
(Image from [here](https://medium.com/@stellasunday095/journey-into-ai-2-understanding-jupyter-notebook-efbed98946e5)).

## The prompts

Notice the `In [ ] :` on the left of this code cell. This is called a *prompt*.  `In` means "Input", which means this is a cell where you type in (e.g. "input") your code.

````python
42
````

When you run the cell, the prompt changes to something like `In [1]:`. Here, `[1]` means this is the first code that the kernel ran.

If there is any output afer you run the cell, you will see it under the cell next to something like `Out [1]:`. Here, 1 is the same number as the number for the input cell.  This is output from the first piece of code that the kernel ran.

## Interrupting, restarting

Sometimes your kernel *stops working* - it will break and will not run the code you want it to.

If this happens you should restart your kernel:

![](https://raw.githubusercontent.com/pxr687/Intro_Comp_Programming/refs/heads/main/images/restart_kernel.png)

**When you restart the kernel, you will need to run all of the cells above the cell you are using.**

## Error messages

Look at the code in the cell below:

- it has a comment (the line which starts with `#`). This is not run as code by the kernel.

- it has some Python code (the line with `print()`). This is run by the kernel.

````python
# This is a comment, it is not code.
print("This is code, it is not a comment!")
````

The reason that we see this output is because the Python code is correct. 

This is the same idea as with a language like English.

You know what the sentence *"we are in a classroom"* means, because it is in correct English.

What about the senetence *"yably har classroom in"*. You do not know what this means, because it has some words that are not English, and also the grammar is not correct.

If you write Python code which has the wrong grammar (e.g. parts of the code are in the wrong order), or you use words Python doees not understand, then you will get an error message:

````python
# This causes an error!
print(This is code, it is not a comment!)
````

Error messages can be confusing! Actually they can be really, really, really, really annoying. BUT they are your friend.

Remember this for the whole course: **you get better at programming by fixing errors. Every error makes you a better programmer.**

When you fix an error, try to understand WHY the error happened.

For now, you can think that the error in the last cell happened because I broke this rule:

> One Python rule is that you have to put text in quotes, `"` or `'`. You will get an error message if you do not.

We will look more at this error later on.

## Getting help in Python

Sometimes, when you write Python code, you will feel like this:

![](https://raw.githubusercontent.com/pxr687/Intro_Comp_Programming/refs/heads/main/images/help_me.png)

However, Python has great built-in ways to get help.

One way is to use the `?` symbol.

This will show you information about something in the code. We have use `print()` a few times now.

`print()` is a Python *function*. A function is Python code that is already written, and it does something useful. In this case, it shows us text.

If we run `?print` we can view the *documentation* for the print function. This tells us what the function does, and how to use it.

````python
# Show the documentation for the `print()` function.
?print
````

You can also also use the `help()` function, to say to Python "help explain this thing to me!".

````python
help(print)
````
