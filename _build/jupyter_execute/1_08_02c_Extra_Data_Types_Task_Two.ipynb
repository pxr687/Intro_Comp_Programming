### Tip 1

### Tip 2
Using [multi-line strings](https://pxr687.github.io/Intro_Comp_Programming/1_08_02_More_On_Strings.html#multi-line-strings) can help you complete this task - you do not have to do this, you can solve the task using one line strings if you want to. Here is a quick demo of multi-line strings:

````python
# A one-line string.
print("This is a string on one line.")
````

````python
# A multi-line string.
print( """
This is a string...

...which goes over lots of...

...lines, I can go over...

...as many lines as I want.

To make a multi-line string...

...you use THREE quotation marks...

...to start and end the string.
""")
````

````python
# You can also store multi-line strings in variables.
my_multi_string = """
This is a string...

...which goes over lots of...

...lines, I can go over...

...as many lines as I want.

To make a multi-line string...

...you use THREE quotation marks...

...to start and end the string.
"""

# `print()` the multi-line string variable.
print(my_multi_string)
````

### Tip 3
You can also look at the [`.replace()` string method](https://pxr687.github.io/Intro_Comp_Programming/1_08_00_Data_Types.html#how-to-avoid-errors-use-type) this will also let you solve this task using less lines of code... Here is a short demo of the `.replace()` string method:

````python
# Make the `my_string` variable.
my_string = "Poto1"

# Show the value of `my_string`.
my_string
````

````python
# Use the `.replace()` string method. 
# Here I replace the character 'o' with the character 'e'.
my_string.replace('o', 'e')
````

````python
# We can use replace lots of times with the same string.
# Here I replace the character 'o' with the character 'e' AND `1` with `r`.
my_string.replace('o', 'e').replace('1', 'r')
````

````python
# We can use `.replace()` lots of times AND save the result in a variable.
my_fixed_string = my_string.replace('o', 'e').replace('1', 'r')

# Show the value of `my_fixed_string`.
my_fixed_string
````
