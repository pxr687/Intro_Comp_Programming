- is `order_total` over 500 baht? Then:

    - if no, show the message `"Shipping fee applies."` and make `shipping_fee` equal 40 baht

```python
order_total = 650

if order_total > 500
    print("Free shipping!")
    shipping_fee = 0
    
else
    print("Shipping fee applies.")
    shipping_fee = 40
```
