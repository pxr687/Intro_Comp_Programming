# Class Rules

## AI

When you do the exercises in this textbook you **should NOT** use AI.

Every line of code should be written by you, from your brain ✅🧠🙂 **NOT** from a robot brain ❌🤖.

If you use AI while you are learning to code for the first time, you will not learn how to code. You will not be able to debug and fix code when it goes wrong. 

Have a look at this [article](https://pxr687.github.io/doc_viewer/AI_assistance.pdf) which explains some of the bad effects of using AI when you are learning new coding skills. It says that:

- using AI when you learn will make you faster to finish coding tasks...

- ...*BUT* you will not understand the code you write, and you will be slower to learn the new skills.

Thinking is *hard*. It takes *effort*. But, like exercising and eating healthily, it is good for you.

Because thinking is hard, it feels good to let an AI think for you. Letting AI think for you, instead of thinking yourself, is called *cognitive offloading*:

![](https://raw.githubusercontent.com/pxr687/Intro_Comp_Programming/refs/heads/main/images/offloading.png)

Cognitive offloading with an AI *feels* good, but it stops you learning. AI can be useful, for example, I used it to make this image. BUT it is bad for you if you use it while you are learning to code...

## Working together in class

When we are in class, you can talk to each other as your code. But you should write every line of code yourself (do not copy from others).

Again, you must use must your own brain, not your friend's brain. As long as you do **NOT** copy any code, you can help each and support each other if you get stuck. 

What is OK:

- Asking a friend for advice on how to fix a bug
