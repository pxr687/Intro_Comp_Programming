## AI

When you do the exercises in this textbook you **should NOT** use AI.

Every line of code should be written by you, from your brain ✅🧠🙂 **NOT** from a robot brain ❌🤖.

If you use AI while you are learning to code for the first time, you will not learn how to code. You will not be able to debug and fix code when it goes wrong. 

Have a look at this [article](https://pxr687.github.io/doc_viewer/AI_assistance.pdf) which explains some of the bad effects of using AI when you are learning new coding skills. It says that:

- using AI when you learn will make you faster to finish coding tasks...

- ...*BUT* you will not understand the code you write, and you will be slower to learn the new skills.

Thinking is *hard*. It takes *effort*. But, like exercising and eating healthily, it is good for you.

Because thinking is hard, it feels good to let an AI think for you. Letting AI think for you, instead of thinking yourself, is called *cognitive offloading* - here are two quotes from the article linked above:

## Computers can make us stupid (even without AI)

```{note}
This section is adapted from [Brett, Rush, Adhikari, Denero and Wagner (2023)](https://odsti.github.io/cfd-textbook/surviving_computers.html). See the [license](https://github.com/pxr687/Intro_Comp_Programming/blob/main/license.md) page for more information.
```
