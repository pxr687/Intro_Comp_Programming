# The `list()` and `dict()` constructors

In Python a *constructor* is a special function that *makes something*. 

Lists and dictionaries have their own constructors, which give us another way to make lists and dictionaries, in addition to the ways that we have seen.

We can make a dictionary using the `dict()` constructor:

```python
dict(name = "Supachai", 
     age = 39,
     country = "Thailand")
```

We can use lists *inside* the constructor to have more than one value per key:

```python
dict(name = ["Supachai", "Cindy"], 
     age = [39, 30],
     country = ["Thailand", "USA"])
```

The `list()` constructor works like this - not we have to use two lots of bracks `(())`:

```python
list(('James', 'Jenny', 'John', 'Jim'))
```
