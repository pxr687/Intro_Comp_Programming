You remember that the following line of code is an *expression* - a recipe that returns a value:

````python
10.50
````

We can see that this is an expression, because Jupyter showed us a text
representation of the value that results: `10.5`.

We know that `10.50` is called a *float literal*.

Now think about this expression:

````python
"Magneto"
````

It is a recipe that returns a value.  In this case the value is Python's
internal representation of the text "Magneto". We know that this is called a *text literal*, or *string literal*.

As we also know, we can store the result of the expression in a variable, using an *assignment
statement*.

We know that the assignment statement has a *variable name* on the left hand side, then an equals sign
`=`, followed by an expression on the right hand side.

````python
villain = "Magneto"

# Show the value.
villain
````

Look at the difference between a *variable name* (no quotes) and the *string
literal* (surrounded by quotes).   A variable name is a name that refers to a
value.  It has to follow the [rules of variable names](1_05_Variables.html#rules-about-variable-names).  A string literal is an
expression, that evaluates to Python's representation of text.

# A whole book as a variable

The strings that we store in variables do not have to be simple, short strings.

To show you the power of string variables, I will now show you now how to get *a whole book* from the internet and store it as a Python variable.

For now, do not worry too much about the code to do this, I will explain what it does:

- first it `import`s a Python library which lets us get data from the internet (from a website address)

- next it makes a variable which has a string of a website address (this variable is called `book_url`

- then it gets a book from that web address, and stores it in a Python variable called `book_text`.

- finally, it prints out the first 5000 lines of the book

The book that we will store as a string variable is [Pride and Prejudice, by Jane Austen](https://en.wikipedia.org/wiki/Pride_and_Prejudice)

````python
# Get the text for Pride and Prejudice.

# Import a library to get data from the internet.
import urllib.request

# Make a string variable which is a website address.
book_url = 'http://www.gutenberg.org/ebooks/42671.txt.utf-8'

# Get the book from the website.
book_text = urllib.request.urlopen(book_url).read().decode()

# Show the first 5000 lines (do not worry about understanding this code for now).
print(book_text[:5000])
````

You can check this is the same as the text you see by opening the URL in your browser: http://www.gutenberg.org/ebooks/42671.txt.utf-8

String data stored in variables like this can be used for lots of different things, including:

- training AI large language models like ChatGPT

- [sentiment analysis](https://www.youtube.com/watch?v=5HQCNAsSO-s), which uses machine learning to learn about the emotions in text. E.g. we could do sentiment analysis with social media posts about a company or product to see if people like or dislike it

- academic research like [Computational Theology](https://www.semanticscholar.org/topic/Computational-theology/3454560) which uses computer science techniques to study religious books
