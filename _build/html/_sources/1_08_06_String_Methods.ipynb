When you have a string, you can make new strings from it using string methods. String methods are *functions* that *work on strings*. These methods are used by putting a dot after the string, then calling the function.

For example, the following method generates an uppercase (all capitals) version of a string.

````python
# Use the `.upper()` string method with the string `loud`.
"loud".upper()
````

````python
# Use the `.upper()` string method with the string `shouting`.
"shouting".upper()
````

We already saw the next string method on an earlier page. It is probably the most important string method - it is called `.replace()`. What it does is *replace* a substring within the string. The `.replace()` method takes two arguments, the text to replace and the text to replace it with.

````python
# Use the `.replace()` string method with the string `"sunny"`.
'sunny'.replace('s', 'f')
````

````python
# Use the `.replace()` string method with the string `"drink"`.
'drink'.replace('dr', 'th')
````

String methods can also be used with variables, if the variable is a string. 

So, for instance, the following code cell makes the word "money" starting from "many" by first making "my", and then doing another replacement after that.

````python
# Our first word, "many".
first_word = "many"

# Use `.replace()` to make our second word ("my").
second_word = first_word.replace('an', '')

# Use `.replace()` to make our third word ("money")
third_word  = second_word.replace('y', 'oney')
third_word
````

Let's break that down to make it clearer what is happening:

````python
# The string we start with.
first_word
````

````python
# Use `.replace()` to make our second word ("my").
second_word = first_word.replace('an', '')
second_word
````

````python
# Replace "in" with "de"
third_word  = second_word.replace('y', 'oney')
third_word
````

Notice that the line `second_word = first_word.replace('an', '')` does **NOT** change the string `first_word`,
which is still "many".
