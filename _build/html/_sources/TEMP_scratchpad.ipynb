```python
# Make a new array (ages).
ages_arr = np.array([17, 26, 22, 18])
```

```python
# Average age.
np.mean(ages_arr)
```

```python
# Total age.
np.sum(ages_arr)
```

```python
ages_arr.sum()
```

```python
# Biggest age.
np.max(ages_arr)
```

```python
ages_arr.max()
```

```python
# Smallest age.
np.min(ages_arr)
```

```python
ages_arr.min()
```
