```python
import webbrowser
import tempfile

roll = np.random.randint(1, 7)

bet = input('Make your bet!:')

if roll == bet:
    emoji = ":)"
else:
    emoji = ":("

message_text = f"""
<!DOCTYPE html>
<html>
<body>
<h1 style="color: green;">
🎲 Dice Roll
</h1>
<p style="color: red;">
The dice rolled a {roll}.
</p>
<p style="color: green;">
You bet it would roll {bet}.
Your bet is {bet == roll} {emoji}
</p>
</body>
</html>
"""

display(HTML(message_text))

# with tempfile.NamedTemporaryFile(
#     mode="w",
#     suffix=".html",
#     delete=False,
#     encoding="utf-8"
# ) as f:
#     f.write(html)
#     filename = f.name

# webbrowser.open_new_tab("file://" + filename)
```
