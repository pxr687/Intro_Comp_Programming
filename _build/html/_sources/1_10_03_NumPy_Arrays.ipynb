NumPy arrays are like the smarter, better-looking version of lists.

```python
my_list[0]
```

```python
my_list[1]
```

```python
my_list[2]
```

```python
my_array = np.array(my_list)

my_array
```

```python
my_array[0]
```

```python
my_array[1]
```

```python
my_array[2]
```

```python
np.sum(my_array)
```
