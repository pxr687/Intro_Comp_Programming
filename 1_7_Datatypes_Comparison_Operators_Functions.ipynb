````python
# Assign a numeric variable.
my_number = 10
````

````python
# Show the value of `my_number`.
my_number
````

````python
type(my_number)
````

````python
# Assign another numeric variable.
my_other_number = 10.0
````

````python
# Show the value of `my_other_number`.
my_other_number
````

````python
type(my_other_number)
````

````python
# Equality comparison operator.
my_number == my_other_number
````

````python
my_extra_number = 200

# Show the value of `my_extra_number`
my_extra_number
````

````python
my_number == my_extra_number
````

````python
is_true = my_number == my_other_number

is_true
````

````python
type(is_true)
````

````python
is_false = my_number == my_extra_number

is_false
````

````python
type(is_false)
````
