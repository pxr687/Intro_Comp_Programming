## Multi-line strings

We can use 3 quotation marks (`'` or `"`) instead of one, so that we can make long strings, without needing loads of `print()` functions!:

````python
print("This is a string on one line.")
````

````python
# Look at the three quotation marks (""") at the start and end of the string.
print("""       
This string...

...can go over...

...as many lines...

...as I want...
""")
````

Note, that if we don't use the `print()` function, we see something different:

````python
"""       
This string...

...can go over...

...as many lines...

...as I want...
"""
````

The `\n` that we see means "new line", it is a special character sequence we can put in a string. The `print()` function knows to display this as a new line.

## F-strings

We have already seen how to put numbers and text together using lots of `+` symbols and the `str` function.

A better way to do this is to use *f-strings*. F-strings are *very* useful. They take some time to get used to, but you will find them very helpful for showing results and debugging your code.

To use a f-string, just put at "f" at the start of your string!:

````python
f"This is an f-string!"
````

Now, inside our f-string, if we put variables, literals or expressions inside curly brackets `{}`, they will become part of the string:

````python
# Make variable.
number_of_students = 17
````

````python
# Look -------------------------------------------------------------------------------------> HERE!
f"This is an f-string! It can show the value of my `number_of_students` variable like this: {number_of_students}"
````

````python
f"We can also put literals/expressions inside our string, like this: {8 * 100}" 
````

Note that using "f" before the string is allowed in Python, because the Python compiler knows it means "this is an f-string! Let me put variables in it!". However, we will get errors if we use other letters, as the Python compiler does not know what these mean:

````python
e"This is not allowed!"
````

## Task 2 - ZenQuotes API and a multi-line F-string

"API" means "Application Programming Interface". APIs are used to let different pieces of software "talk" to each other by sending data. 

The code cell below "talks" to the [ZenQuotes API](https://docs.zenquotes.io/category/guides/). It sends a request for data (e.g. it asks ZenQuotes "please send me a quote!" over the internet) and receives data back. We will use this API again later in the course. 

At the moment, it gets one `quote` and one `author`, both as Python strings. The comments explain what the code is doing. Read the comments and then run the cell:

````python
# DO NOT CHANGE THIS CODE. READ THE COMMENTS AND THEN RUN THE CELL.

# Importing some built-in Python libraries for sending
# requests to websites (`urllib`) and also for processing
# json (`json`) files.
import urllib.request
import json

# Store the URL of the ZenQuotes API, as a Python string.
api_web_address = "https://zenquotes.io/api/random"

# Use `urllib` to request data from ZenQuotes.
with urllib.request.urlopen(api_web_address) as response:
    
    # This part is quite hard to understand. The first line
    # converts the data into a Python list (we will learn
    # more about these later)
    full = json.loads(response.read().decode('utf-8'))
    
    # Then these lines "grab" the quote and the author
    # from the list (it is actually a dictionary inside
    # a list, but we will talk about this in a later class).
    quote = full[0]['q']
    author = full[0]['a']
    
# Then we show the quote, and the author.
print(quote)
print(author)
````

If you want a different quote, just re-run the cell. However, if you do this too much within a minute, you will hit the API [rate limit](https://blog.postman.com/what-is-api-rate-limiting/) which stops users from overloading the server with too many requests. You'll see an error if this happens, just wait a minute or two and you'll be able to send a new request.
